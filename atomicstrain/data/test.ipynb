```python
def comma_to_float(s):
    return float(s.replace(',', '.'))

data = np.loadtxt("results/data/avg_shear_strain.txt", usecols=(1, 4), converters={1: comma_to_float, 4: comma_to_float}, dtype=float)
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(data[:, 0], data[:, 1], linestyle='-', color='black')
ax.set_xlabel('Residue', fontsize=14)
ax.set_ylabel('Strain', fontsize=14)
ax.set_title('Strain vs Distance', fontsize=16)
plt.show()
```
